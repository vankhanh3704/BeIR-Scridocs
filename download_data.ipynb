```python
import os
from datasets import load_dataset
import pandas as pd

# Khai báo thu muc
RAW_DIR = "data/raw"
PREPROCESSED_DIR = "data/processed"
MODELS = "models"
for d in [RAW_DIR, MODELS, PREPROCESSED_DIR]:
    os.makedirs(d, exist_ok=True)

# Khai bao duong dan file
CORPUS_PATH = f"{RAW_DIR}/corpus.parquet"
QUERIES_PATH = f"{RAW_DIR}/queries.parquet"
QRELS_PATH = f"{RAW_DIR}/qrels.parquet"
```

```python
# Kiem tra va tai du lieu
if not (os.path.exists(CORPUS_PATH) and os.path.exists(QUERIES_PATH) and os.path.exists(QRELS_PATH)):
    print("Downloading raw SCIDOCS data from Hugging Face...")
    
    hf_corpus = load_dataset("BeIR/scidocs", "corpus")["corpus"]
    hf_queries = load_dataset("BeIR/scidocs", "queries")["queries"]
    hf_qrels = load_dataset("BeIR/scidocs-qrels")["test"]

    hf_corpus.to_parquet(CORPUS_PATH)
    hf_queries.to_parquet(QUERIES_PATH)
    hf_qrels.to_parquet(QRELS_PATH)
    
    print("Download complete. Files saved to data/raw.")
else:
    print("Files already exist. Skipping download.")

# Kiem tra dung luong file
print("\nDirectory contents:")
for f in os.listdir(RAW_DIR):
    if f.endswith(".parquet"):
        size_mb = os.path.getsize(f"{RAW_DIR}/{f}") / (1024 * 1024)
        print(f"  - {f:<20} {size_mb:.2f} MB")
```

```python
# Doc cac file du lieu tho
corpus_raw = pd.read_parquet(CORPUS_PATH)
queries_raw = pd.read_parquet(QUERIES_PATH)
qrels_raw = pd.read_parquet(QRELS_PATH)

print("=== THONG TIN CORPUS ===")
print(f"So luong: {corpus_raw.shape[0]:,} tai lieu")
print(f"Cac cot: {list(corpus_raw.columns)}")
print(corpus_raw.head(2))

print("\n=== THONG TIN QUERIES ===")
print(f"So luong: {queries_raw.shape[0]:,} cau hoi")
print(f"Cac cot: {list(queries_raw.columns)}")
print(queries_raw.head(2))

print("\n=== THONG TIN QRELS (DAP AN) ===")
print(f"So luong: {qrels_raw.shape[0]:,} dong")
print(f"Cac cot: {list(qrels_raw.columns)}")
print(qrels_raw.head(2))
```
